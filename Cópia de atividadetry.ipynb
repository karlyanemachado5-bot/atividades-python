#Modelos de tratamento python resolvidos

Questao 1 - positivo, negativo ou zero

```python
while True:
  try:
    numero = int(input("Digite um numero: "))
    break
  except ValueError:
    print("Digite um numero inteiro valido.")

if numero > 0:
  print("Positivo")
elif numero < 0:
  print("Negativo")
else:
  print("Zero")
```

Questao 2 - maior entre dois numeros


```python
valores = []
for i in range(2):
  while True:
      try:
         n = int(input(f"Digite o {i+1}º numero: "))
         valores.append(n)
         break
      except ValueError:
        print("Informe um inteiro valido.")

a, b = valores
if a > b:
  print("Maior:", a)
elif b > a:
  print("Maior:", b)
else:
  print("Os numeros sao iguais.")
```

Questao 3 - somando cinco numeros


```python
soma = 0
for i in range(5):
  while True:
      try:
          numero = int(input(f"{i+1}º numero: "))
          soma += numero
          break
      except ValueError:
          print("Entrada invalida. Tente novamente.")

print("Soma:", soma)
```

Questao 4 - par ou impar


```python
while True:
    try:
        numero = int(input("Digite um inteiro: "))
        break
    except ValueError:
        print("Use apenas um número inteiro.")

if numero % 2 == 0:
    print("Par")
else:
    print("Ímpar")
```

Questao 5 - situação do aluno

```python
notas = []

for i in range(2):
    while True:
        try:
            nota = float(input(f"Digite a {i+1}ª nota: "))

            if 0 <= nota <= 10:
                notas.append(nota)
                break

            print("A nota deve estar entre 0 e 10.")

        except ValueError:
            print("Digite uma nota numérica válida.")

media = sum(notas) / 2

if media >= 7:
    print("Aprovado")
elif media >= 5:
    print("Recuperação")
else:
    print("Reprovado")

print("Média:", media)
```

Questao 6 - maior entre números

```python
valores = []

for i in range(3):
    while True:
        try:
            numero = int(input(f"Digite o {i+1}º número: "))
            valores.append(numero)
            break
        except ValueError:
            print("Digite um número inteiro válido.")

maior = valores[0]

for numero in valores:
    if numero > maior:
        maior = numero

print("Maior número:", maior)
```

Questao 8 - soma de 1 até N

```python
while True:
    try:
        n = int(input("Digite um número positivo: "))

        if n > 0:
            break

        print("Digite um número maior que zero.")

    except ValueError:
        print("Entrada inválida.")

soma = 0

for i in range(1, n + 1):
    soma += i

print("Soma:", soma)
```

Questao 9 - contando números pares

```python
pares = 0

for i in range(10):
    while True:
        try:
            numero = int(input(f"Digite o {i+1}º número: "))
            break
        except ValueError:
            print("Digite um número inteiro válido.")

    if numero % 2 == 0:
        pares += 1

print("Quantidade de números pares:", pares)
```

Questao 10 - soma apenas dos positivos

```python
soma = 0

for i in range(8):
    while True:
        try:
            numero = int(input(f"Digite o {i+1}º número: "))
            break
        except ValueError:
            print("Digite um número inteiro válido.")

    if numero > 0:
        soma += numero

print("Soma dos positivos:", soma)
```

Questao 11 - maior valor digitado

```python
maior = None

for i in range(6):
    while True:
        try:
            numero = int(input(f"Digite o {i+1}º número: "))
            break
        except ValueError:
            print("Digite um número inteiro válido.")

    if maior is None or numero > maior:
        maior = numero

print("Maior valor:", maior)
```

Questao 12 - tabuada

```python
while True:
    try:
        numero = int(input("Digite um número inteiro: "))
        break
    except ValueError:
        print("Digite um número inteiro válido.")

for i in range(1, 11):
    print(numero, "x", i, "=", numero * i)
```

Questao 13 - tentativas de senha

```python
senha_correta = "1234"

while True:
    senha = input("Digite a senha: ")

    if senha == senha_correta:
        print("Senha correta! Acesso permitido.")
        break
    else:
        print("Senha incorreta. Tente novamente.")
```

Questao 14 - número secreto

```python
secreto = 7
tentativas = 0

while True:
    try:
        numero = int(input("Tente adivinhar o número: "))
        tentativas += 1

        if numero == secreto:
            print("Parabéns! Você acertou.")
            break
        elif numero < secreto:
            print("O número secreto é maior.")
        else:
            print("O número secreto é menor.")

    except ValueError:
        print("Digite um número inteiro válido.")

print("Tentativas:", tentativas)
```

Questao 15 - analisando uma sequência

```python
quantidade = 0
positivos = 0
negativos = 0
soma = 0

while True:
    try:
        numero = int(input("Digite um número (0 para encerrar): "))

        if numero == 0:
            break

        quantidade += 1
        soma += numero

        if numero > 0:
            positivos += 1
        else:
            negativos += 1

    except ValueError:
        print("Entrada inválida. Tente novamente.")

print("Quantidade:", quantidade)
print("Positivos:", positivos)
print("Negativos:", negativos)
print("Soma:", soma)
```
