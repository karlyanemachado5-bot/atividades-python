```python
horas = int(input("Digite a quantidade de horas: "))

if horas <= 2:
    total = 8.0
elif horas <= 5:
    total = 8.0 + (horas - 2) * 3.0
else:
    total = 17.0 + (horas - 5) * 2.0

print(f"Valor a pagar: R$ {total:.2f}")
```

```python
repetido = False
anterior = int(input("Digite o 1º número: "))

for i in range(2, 9):
    num = int(input(f"Digite o {i}º número: "))
    if num == anterior:
        repetido = True
    anterior = num

if repetido:
    print("Existem valores consecutivos iguais.")
else:
    print("Não existem valores consecutivos iguais.")
```

```python
saldo = 500.0

for i in range(1, 7):
    tipo = int(input("Operação (1-Depósito, 2-Saque): "))
    valor = float(input("Valor: R$ "))

    if tipo == 1:
        saldo += valor
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente")

print(f"Saldo final: R$ {saldo:.2f}")
```

```python
num = int(input("Digite um inteiro maior que 1: "))

eh_primo = True
for i in range(2, num):
    if num % i == 0:
        eh_primo = False
        break

if eh_primo:
    print(f"{num} é primo.")
else:
    print(f"{num} não é primo.")
```

```python
n = int(input("Digite um número inteiro positivo: "))
qtd_divisores = 0

print("Divisores:")
for i in range(1, n + 1):
    if n % i == 0:
        print(i, end=" ")
        qtd_divisores += 1

print(f"\nTotal de divisores: {qtd_divisores}")
```

```python
a = int(input("Digite A: "))
b = int(input("Digite B: "))

qtd_primos = 0
print("Primos encontrados:")

for num in range(a, b + 1):
    if num > 1:
        eh_primo = True
        for i in range(2, num):
            if num % i == 0:
                eh_primo = False
                break
        if eh_primo:
            print(num, end=" ")
            qtd_primos += 1

print(f"\nTotal de primos: {qtd_primos}")
```

```python
secreto = 37
tentativas = 5
acertou = False

while tentativas > 0 and not acertou:
    palpite = int(input("Digite seu palpite: "))
    tentativas -= 1

    if palpite == secreto:
        print("Parabéns! Você acertou!")
        acertou = True
    elif palpite < secreto:
        print("O número secreto é MAIOR.")
    else:
        print("O número secreto é MENOR.")

if not acertou:
    print(f"Fim de jogo! O número secreto era {secreto}.")
```

```python
saldo = 200.00
qtd_entradas = 0
qtd_retiradas = 0
qtd_recusadas = 0
total_entrada = 0.00
total_saida = 0.00

while True:
    print("\n--- MENU ---")
    print("1 - Entrada de dinheiro")
    print("2 - Retirada de dinheiro")
    print("0 - Encerrar")
    opcao = int(input("Escolha uma operação: "))

    if opcao == 0:
        break
    elif opcao == 1:
        valor = float(input("Valor da entrada: R$ "))
        saldo += valor
        total_entrada += valor
        qtd_entradas += 1
    elif opcao == 2:
        valor = float(input("Valor da retirada: R$ "))
        if valor <= saldo:
            saldo -= valor
            total_saida += valor
            qtd_retiradas += 1
        else:
            print("Saldo insuficiente — retirada recusada!")
            qtd_recusadas += 1
    else:
        print("Opção inválida!")

print("\n" + "="*35)
print(f"Saldo final:      R$ {saldo:.2f}")
print(f"Entradas:         {qtd_entradas} | Total: R$ {total_entrada:.2f}")
print(f"Retiradas:        {qtd_retiradas} | Total: R$ {total_saida:.2f}")
print(f"Retiradas recusadas: {qtd_recusadas}")
print("="*35)
```
