Questao 4 - estacionamento

```python
while True:
    try:
        horas = int(input("Digite a quantidade de horas: "))

        if horas > 0:
            break
        else:
            print("Digite uma quantidade válida.")

    except ValueError:
        print("Entrada inválida. Digite um número inteiro.")

if horas <= 2:
    valor = 8
elif horas <= 5:
    valor = 8 + (horas - 2) * 3
else:
    valor = 17 + (horas - 5) * 2

print("Valor a pagar: R$", valor)
```

Questao 5 - máquina de venda de ingresso

```python
total = 0

while total < 25:
    try:
        dinheiro = int(input("Digite o valor inserido: "))

        if dinheiro > 0:
            total += dinheiro
        else:
            print("Digite um valor positivo.")

    except ValueError:
        print("Valor inválido. Tente novamente.")

print("Total inserido: R$", total)
print("Troco: R$", total - 25)
```

Questao 6 - maior valor e sua posição

```python
maior = None
posicao = 0

for i in range(1, 9):
    while True:
        try:
            numero = int(input("Digite um número: "))
            break
        except ValueError:
            print("Digite um número inteiro válido.")

    if maior is None or numero > maior:
        maior = numero
        posicao = i

print("Maior valor:", maior)
print("Primeira posição:", posicao)
```

Questao 7 - segundo maior valor

```python
numeros = []
while len(numeros) < 8:
    try:
        numero = int(input("Digite um número: "))

        if numero not in numeros:
            numeros.append(numero)
        else:
            print("Número repetido. Digite outro.")

    except ValueError:
        print("Digite um número inteiro válido.")

maior = segundo = None

for numero in numeros:
    if maior is None or numero > maior:
        segundo = maior
        maior = numero
    elif segundo is None or numero > segundo:
        segundo = numero

print("Maior:", maior)
print("Segundo maior:", segundo)
```

Questao 8 - detectando repetição consecutiva

```python
anterior = None
repetiu = False

for i in range(8):
    while True:
        try:
            numero = int(input("Digite um número: "))
            break
        except ValueError:
            print("Digite um número inteiro válido.")

    if numero == anterior:
        repetiu = True

    anterior = numero

if repetiu:
    print("Existem valores consecutivos iguais.")
else:
    print("Não existem valores consecutivos iguais.")
```

Questao 9 - controle de saldo

```python


saldo = 500

for i in range(6):
    while True:
        try:
            tipo = int(input("Digite 1 para depósito ou 2 para saque: "))

            if tipo == 1 or tipo == 2:
                break
            else:
                print("Tipo inválido.")

        except ValueError:
            print("Digite um número válido.")

    while True:
        try:
            valor = float(input("Digite o valor: "))

            if valor > 0:
                break
            else:
                print("Digite um valor positivo.")

        except ValueError:
            print("Valor inválido.")

    if tipo == 1:
        saldo += valor
    elif valor <= saldo:
        saldo -= valor
    else:
        print("Saldo insuficiente.")

print("Saldo final: R$", saldo)
```

Questao 10 - Caixa eletrônico simplificado

```python
while True:
    try:
        valor = int(input("Digite o valor do saque: "))

        if valor > 0:
            break
        else:
            print("Digite um valor positivo.")

    except ValueError:
        print("Entrada inválida.")

cedulas = [100, 50, 20, 10, 5, 2, 1]

for cedula in cedulas:
    quantidade = valor // cedula
    valor = valor % cedula

    if quantidade > 0:
        print(quantidade, "cédula(s) de R$", cedula)
```

Questao 11 - número primo

```python
while True:
    try:
        numero = int(input("Digite um número maior que 1: "))

        if numero > 1:
            break
        else:
            print("O número deve ser maior que 1.")

    except ValueError:
        print("Digite um número inteiro válido.")

primo = True

for i in range(2, numero):
    if numero % i == 0:
        primo = False
        break

if primo:
    print("O número é primo.")
else:
    print("O número não é primo.")
```

Questao 12 - divisores de um número

```python
while True:
    try:
        n = int(input("Digite um número positivo: "))

        if n > 0:
            break
        else:
            print("Digite um número maior que zero.")

    except ValueError:
        print("Entrada inválida.")

quantidade = 0

print("Divisores:")

for i in range(1, n + 1):
    if n % i == 0:
        print(i)
        quantidade += 1

print("Quantidade de divisores:", quantidade)
```

Questao 13 - números primos em um intervalo

```python
while True:
    try:
        a = int(input("Digite o primeiro número: "))
        b = int(input("Digite o segundo número: "))

        if a > 0 and b > 0 and a < b:
            break
        else:
            print("Digite números positivos e A menor que B.")

    except ValueError:
        print("Digite números inteiros válidos.")

quantidade = 0

print("Números primos:")

for numero in range(a, b + 1):
    if numero > 1:
        primo = True

        for i in range(2, numero):
            if numero % i == 0:
                primo = False
                break

        if primo:
            print(numero)
            quantidade += 1

print("Quantidade de primos:", quantidade)
```

Questao 14 - jogo com limite de tentativas

```python
secreto = 37
tentativas = 0

while tentativas < 5:
    try:
        palpite = int(input("Tente adivinhar o número: "))

        tentativas += 1

        if palpite == secreto:
            print("Parabéns! Você acertou.")
            break
        elif palpite < secreto:
            print("O número secreto é maior.")
        else:
            print("O número secreto é menor.")

    except ValueError:
        print("Digite um número inteiro válido.")

if tentativas == 5 and palpite != secreto:
    print("Suas tentativas acabaram. O número era 37.")
```

```python
saldo = 200
entradas = 0
retiradas = 0
recusadas = 0
total_entrou = 0
total_saiu = 0

while True:
    try:
        codigo = int(input("Digite 1 para entrada, 2 para retirada ou 0 para sair: "))

        if codigo == 0:
            break
        elif codigo != 1 and codigo != 2:
            print("Código inválido.")
            continue

    except ValueError:
        print("Digite um código válido.")
        continue

    while True:
        try:
            valor = float(input("Digite o valor da operação: "))

            if valor > 0:
                break
            else:
                print("Digite um valor positivo.")

        except ValueError:
            print("Valor inválido.")

    if codigo == 1:
        saldo += valor
        entradas += 1
        total_entrou += valor

    elif codigo == 2:
        if valor <= saldo:
            saldo -= valor
            retiradas += 1
            total_saiu += valor
        else:
            print("Saldo insuficiente.")
            recusadas += 1

print("Saldo final: R$", saldo)
print("Quantidade de entradas:", entradas)
print("Retiradas realizadas:", retiradas)
print("Retiradas recusadas:", recusadas)
print("Total que entrou: R$", total_entrou)
print("Total que saiu: R$", total_saiu)
```
