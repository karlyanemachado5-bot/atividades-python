```python
#Questao 4
def mostrar_valores(obrigatorio, *args):
    print("Obrigatório:", obrigatorio)
    print("Adicionais:", args)

# Apenas 1 elemento
mostrar_valores("Python")

# 4 elementos
mostrar_valores("Python", "Java", "C++", "JavaScript")
```

```python
#Questao 7
#F=\frac{9}{5}C+32
temperaturas_c = [39.2, 36.5, 37.3, 37.8]

converter = lambda c: (c * 9/5) + 32

temperaturas_f = list(map(converter, temperaturas_c))

print(temperaturas_f)
```
