1. Guardando e analisando notas

```python
notas = []
for i in range(5):
    nota = float(input("Digite a nota: "))
    notas.append(nota)

soma = 0
for nota in notas:
    soma += nota

media = soma / len(notas)

acima_ou_igual = 0
for nota in notas:
    if nota >= media:
        acima_ou_igual += 1

print("Notas:", notas)
print("Média:", media)
print("Notas >= média:", acima_ou_igual)
```

4. Lista de números: pares e ímpares

```python
numeros = []
pares = []
impares = []

for i in range(10):
    numero = int(input(f"Digite o {i+1}º número: "))
    numeros.append(numero)

for numero in numeros:
    if numero % 2 == 0:
        pares.append(numero)
    else:
        impares.append(numero)

print("Lista original:", numeros)
print("Pares:", pares)
print("Ímpares:", impares)
```

5. Maior e menor da lista

```python
numeros = []

for i in range(8):
    numero = int(input(f"Digite o {i+1}º número: "))
    numeros.append(numero)

maior = numeros[0]
menor = numeros[0]

pos_maior = 0
pos_menor = 0

for i in range(1, len(numeros)):
    if numeros[i] > maior:
        maior = numeros[i]
        pos_maior = i

    if numeros[i] < menor:
        menor = numeros[i]
        pos_menor = i

print("Lista:", numeros)
print("Maior:", maior, "Posição:", pos_maior)
print("Menor:", menor, "Posição:", pos_menor)
```

6. Nomes sem repetição

```python
nomes = []

while True:
    nome = input("Digite um nome ou FIM para encerrar: ")

    if nome.upper() == "FIM":
        break

    if nome not in nomes:
        nomes.append(nome)

print("Nomes cadastrados:", nomes)
print("Quantidade de nomes diferentes:", len(nomes))
```

8. Estatísticas de uma tupla

```python
numeros = []

for i in range(6):
    numero = int(input(f"Digite o {i+1}º número: "))
    numeros.append(numero)

numeros = tuple(numeros)

quantidade_5 = 0
posicao_5 = -1
pares = 0

for i in range(len(numeros)):
    if numeros[i] == 5:
        quantidade_5 += 1

        if posicao_5 == -1:
            posicao_5 = i

    if numeros[i] % 2 == 0:
        pares += 1

print("Tupla:", numeros)
print("Quantidade de vezes que o 5 aparece:", quantidade_5)

if posicao_5 != -1:
    print("Posição da primeira ocorrência do 5:", posicao_5)
else:
    print("O número 5 não aparece.")

print("Quantidade de valores pares:", pares)
```

11. Frequência das palavras

```python
frequencia = {}

for i in range(8):
    palavra = input(f"Digite a {i+1}ª palavra: ")

    if palavra in frequencia:
        frequencia[palavra] += 1
    else:
        frequencia[palavra] = 1

print("\nFrequência das palavras:")

for palavra, quantidade in frequencia.items():
    print(palavra, ":", quantidade)
```

12. Estoque e atualização

```python
estoque = {
    "Arroz": 10,
    "Feijão": 8,
    "Leite": 15,
    "Açúcar": 12
}

produto = input("Digite o produto vendido: ")
quantidade = int(input("Digite a quantidade vendida: "))

if produto in estoque:
    if estoque[produto] >= quantidade:
        estoque[produto] -= quantidade
        print("Venda realizada com sucesso!")
    else:
        print("Estoque insuficiente!")
else:
    print("Produto não cadastrado.")

print("Estoque final:")
print(estoque)
```

14. Catálogo com código imutável

```python
produtos = [
    (101, "Caderno", 20.00),
    (102, "Caneta", 3.50),
    (103, "Mochila", 80.00),
    (104, "Lápis", 2.00),
    (105, "Borracha", 1.50)
]

codigo = int(input("Digite o código do produto: "))

encontrado = False

for produto in produtos:
    if produto[0] == codigo:
        print("Nome:", produto[1])
        print("Preço: R$", produto[2])
        encontrado = True
        break

if not encontrado:
    print("Código não encontrado.")
```

15. Mini sistema de estoque

```python
estoque = [
    {"codigo": 101, "nome": "Caderno", "quantidade": 10, "preco": 20.00},
    {"codigo": 102, "nome": "Caneta", "quantidade": 30, "preco": 3.50},
    {"codigo": 103, "nome": "Mochila", "quantidade": 5, "preco": 80.00},
    {"codigo": 104, "nome": "Lápis", "quantidade": 20, "preco": 2.00},
    {"codigo": 105, "nome": "Borracha", "quantidade": 15, "preco": 1.50}
]

valor_total = 0
maior_valor = 0
item_maior_valor = None

print("=== ESTOQUE ===")

for item in estoque:
    print(
        item["codigo"],
        "-", item["nome"],
        "- Quantidade:", item["quantidade"],
        "- Preço: R$", item["preco"]
    )

    valor_item = item["quantidade"] * item["preco"]
    valor_total += valor_item

    if valor_item > maior_valor:
        maior_valor = valor_item
        item_maior_valor = item

quantidade_itens = len(estoque)

resumo = (quantidade_itens, valor_total)

print("\nValor total do estoque: R$", valor_total)
print("Item de maior valor em estoque:", item_maior_valor["nome"])
print("Valor desse item: R$", maior_valor)
print("Resumo:", resumo)
```
